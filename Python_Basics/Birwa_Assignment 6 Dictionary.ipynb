```python

# ==============================================================================
# GROUP 5: AGGREGATIONS AND ADVANCED ANALYSIS (QUESTIONS 21 - 30)
# ==============================================================================

# 21. Find the project with the highest funding.
highest_funding = 0
top_project = ""
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "labs" in dept_data:
                for lab_data in dept_data["labs"].values():
                    for proj, p_data in lab_data["research_projects"].items():
                        if p_data["funding"] > highest_funding:
                            highest_funding = p_data["funding"]
                            top_project = proj
print(f"Highest Funded Project: {top_project} (${highest_funding})")


# 22. Find total funding across all universities.
total_funding = 0
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "labs" in dept_data:
                for lab_data in dept_data["labs"].values():
                    for p_data in lab_data["research_projects"].values():
                        total_funding += p_data["funding"]
print(total_funding)


# 23. Find the university with the best ranking.
best_rank = float('inf')
best_uni = ""
for uni, uni_data in university_system["universities"].items():
    if uni_data["ranking"] < best_rank:
        best_rank = uni_data["ranking"]
        best_uni = uni
print(best_uni)


# 24. Print all completed projects.
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "labs" in dept_data:
                for lab_data in dept_data["labs"].values():
                    for proj, p_data in lab_data["research_projects"].items():
                        if p_data["status"] == "completed":
                            print(proj)


# 25. Count total number of researchers in GTU_India.
india_researchers = set()
cs_labs = university_system["universities"]["GTU_India"]["faculties"]["engineering"]["departments"]["computer_science"]["labs"]
for lab_data in cs_labs.values():
    for p_data in lab_data["research_projects"].values():
        for researcher in p_data["researchers"]:
            india_researchers.add(researcher)
print(len(india_researchers))


# 26. Find which lab has the maximum funding.
lab_funding = {}
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "labs" in dept_data:
                for lab_name, lab_data in dept_data["labs"].items():
                    current_funding = sum(p_data["funding"] for p_data in lab_data["research_projects"].values())
                    lab_funding[lab_name] = current_funding
top_lab = max(lab_funding, key=lab_funding.get)
print(f"Top Lab: {top_lab} with {lab_funding[top_lab]}")


# 27. Print projects grouped by status (ongoing/completed).
grouped_projects = {"ongoing": [], "completed": []}
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "labs" in dept_data:
                for lab_data in dept_data["labs"].values():
                    for proj, p_data in lab_data["research_projects"].items():
                        grouped_projects[p_data["status"]].append(proj)
print(grouped_projects)


# 28. Print all courses with more than 100 students enrolled.
for uni, uni_data in university_system["universities"].items():
    for fac_data in uni_data["faculties"].values():
        for dept_data in fac_data["departments"].values():
            if "courses" in dept_data:
                for course, c_data in dept_data["courses"].items():
                    if c_data["students_enrolled"] > 100:
                        print(course)


#29. Find total number of faculties in the entire system.
total_faculties = 0
for uni, uni_data in university_system["universities"].items():
    total_faculties += len(uni_data["faculties"])

print(total_faculties)

#30. Write a function that searches for a project name and prints its full details.
def search_project(project_name):
    for uni, uni_data in university_system["universities"].items():
        for fac_data in uni_data["faculties"].values():
            for dept_data in fac_data["departments"].values():
                if "labs" in dept_data:
                    for lab_data in dept_data["labs"].values():
                        if project_name in lab_data["research_projects"]:
                            return lab_data["research_projects"][project_name]
                        return "Project not found"

print(f"DeepVision Project Details: {search_project('DeepVision')}")
```
