```python
# Q8. Find all employees whose email address is missing
print("Employees with missing email addresses:")
for emp_id, info in employees.items():
    if info["Email"] is None:
        print(f"- {info['Name']} (ID: {emp_id})")

# Q9. Find all employees whose phone number is missing
print("\nEmployees with missing phone numbers:")
for emp_id, info in employees.items():
    if info["Phone"] is None:
        print(f"- {info['Name']} (ID: {emp_id})")

# Q10. Count incomplete records (missing email OR phone)
incomplete_count = sum(1 for info in employees.values() if info["Email"] is None or info["Phone"] is None)
print(f"\nTotal employees with incomplete records: {incomplete_count}")

# Q11. Replace all missing email addresses with 'Not Available'
for info in employees.values():
    if info["Email"] is None:
        info["Email"] = "Not Available"
```

```python
# Q17. Find the employee with the highest salary
highest_emp = max(employees.values(), key=lambda x: x["Salary"])
print(f"Highest Salary: {highest_emp['Name']} earning ₹{highest_emp['Salary']}")

# Q18. Find the employee with the lowest salary
lowest_emp = min(employees.values(), key=lambda x: x["Salary"])
print(f"Lowest Salary: {lowest_emp['Name']} earning ₹{lowest_emp['Salary']}")

# Q19. Calculate the average salary of all employees
all_salaries = [info["Salary"] for info in employees.values()]
avg_salary = sum(all_salaries) / len(all_salaries)
print(f"Average Salary: ₹{avg_salary:.2f}")

# Q20. Display employees earning more than ₹100,000
print("\nEmployees earning more than ₹100,000:")
for info in employees.values():
    if info["Salary"] > 100000:
        print(f"- {info['Name']} (₹{info['Salary']})")

# Q21. Display employees earning less than ₹50,000
print("\nEmployees earning less than ₹50,000:")
for info in employees.values():
    if info["Salary"] < 50000:
        print(f"- {info['Name']} (₹{info['Salary']})")
```

```python
# Q22. Count how many employees belong to each department
dept_counts = {}
for info in employees.values():
    dept = info["Department"]
    dept_counts[dept] = dept_counts.get(dept, 0) + 1

print("Department Breakdown:")
for dept, count in dept_counts.items():
    print(f"{dept} : {count}")

# Q23. Count how many employees are managed by each manager
manager_counts = {}
for info in employees.values():
    manager = info["Manager"]
    manager_counts[manager] = manager_counts.get(manager, 0) + 1

print("\nManager Assignment Breakdown:")
for manager, count in manager_counts.items():
    print(f"{manager} : {count}")

# Q24. Find which department has the largest number of employees
largest_dept = max(dept_counts, key=dept_counts.get)
print(f"\nDepartment with the largest workforce: {largest_dept} ({dept_counts[largest_dept]} employees)")
```

```python
# Extract all valid numeric salary entries
salaries = [info["Salary"] for info in employees.values()]

# Calculate requested metrics
total_employees = len(employees)
missing_email = sum(1 for info in employees.values() if info["Email"] is None)
missing_phone = sum(1 for info in employees.values() if info["Phone"] is None)
avg_salary = sum(salaries) / len(salaries) if salaries else 0
highest_salary = max(salaries) if salaries else 0
lowest_salary = min(salaries) if salaries else 0

# Print the report format
print("--- EMPLOYEE DATABASE REPORT ---")
print(f"Total Employees: {total_employees}")
print(f"Employees with Missing Email: {missing_email}")
print(f"Employees with Missing Phone: {missing_phone}")
print(f"Average Salary: ₹{avg_salary:.2f}")
print(f"Highest Salary: ₹{highest_salary}")
print(f"Lowest Salary: ₹{lowest_salary}")
print("-" * 32)
```

```python
# Create a dictionary to group employees by department
dept_groups = {}

for info in employees.values():
    dept = info["Department"]
    if dept not in dept_groups:
        dept_groups[dept] = []
    dept_groups[dept].append(info["Name"])

# Print group results with headings
for dept, names in dept_groups.items():
    print(dept)
    print("-" * len(dept))
    for name in names:
        print(name)
    print()  # Empty row separator
```

```python
# Sort items using a lambda function targeted at the Salary value
sorted_employees = sorted(employees.values(), key=lambda x: x["Salary"], reverse=True)

print("Employees Ranked by Salary (Highest to Lowest):")
for info in sorted_employees:
    print(f"- {info['Name']}: ₹{info['Salary']}")
```

```python
# 1. Count how many employees each manager has
manager_counts = {}
for info in employees.values():
    mgr = info["Manager"]
    manager_counts[mgr] = manager_counts.get(mgr, 0) + 1

# 2. Filter and print results
print("Managers supervising more than 8 employees:")
for mgr, count in manager_counts.items():
    if count > 8:
        print(f"- {mgr} (Supervises {count} employees)")
```

```python
# 1. Gather all salaries per department
dept_salaries = {}
for info in employees.values():
    dept = info["Department"]
    if dept not in dept_salaries:
        dept_salaries[dept] = []
    dept_salaries[dept].append(info["Salary"])

# 2. Find individual averages and pick the max value
dept_averages = {}
for dept, salary_list in dept_salaries.items():
    dept_averages[dept] = sum(salary_list) / len(salary_list)

highest_avg_dept = max(dept_averages, key=dept_averages.get)
print(f"Highest Average Salary Department: {highest_avg_dept} (Average: ₹{dept_averages[highest_avg_dept]:.2f})")
```

```python
print("Employees missing BOTH email and phone records:")
found_any = False
for info in employees.values():
    if info["Email"] is None and info["Phone"] is None:
        print(f"- {info['Name']}")
        found_any = True

if not found_any:
    print("- No employees are missing both fields.")
```

```python
# Trace duplicates by recording seen values
seen_phones = set()
duplicate_phones = set()

for info in employees.values():
    phone = info["Phone"]
    if phone is not None:
        if phone in seen_phones:
            duplicate_phones.add(phone)
        seen_phones.add(phone)

if duplicate_phones:
    print(f"Duplicate phone numbers found: {duplicate_phones}")
else:
    print("No duplicate phone numbers exist in the database.")
```

```python
print("Employees with a 'company.com' email address:")
for info in employees.values():
    # Safely check string endings while filtering out None values
    if info["Email"] is not None and info["Email"].endswith("@company.com"):
        print(f"- {info['Name']} ({info['Email']})")
```
