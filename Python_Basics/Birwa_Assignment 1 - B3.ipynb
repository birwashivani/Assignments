```python
# Section G –  Questions

Write a single slicing statement to extract:

Every alternate matrix.
Every alternate row from all matrices.
Every alternate column from all matrices.
The first column from every matrix.
The last column from every matrix.
The middle row from every matrix.
The middle column from every matrix.
Reverse the order of matrices.
Reverse the rows inside every matrix.
Reverse the columns inside every matrix.
```

```python
#print(arr)
print("Every alternate matrix \n", arr[::2])
print("Every alternate row from all matrices \n", arr[:, ::2])
print("Every alternate column from all matrices \n", arr[:, :, ::2])
print("The first column from every matrix \n", arr[:, :, 0:1])
print("The last column from every matrix \n", arr[:, :, 4:5])
print("The middle row from every matrix \n", arr[:, 1:2])
print("The middle column from every matrix \n", arr[:, :, 2:3])
print("Reverse the order of matrices \n", arr[::-1, :, :])
print("Reverse the rows inside every matrix \n", arr[:, ::-1, :])
print("Reverse the columns inside every matrix \n", arr[:, :, ::-1])
```
